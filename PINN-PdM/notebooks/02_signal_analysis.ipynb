# 02 — Signal Analysis

Raw waveform, FFT and STFT of a synthetic fault window with characteristic-frequency overlays.

```python
import sys, os
sys.path.insert(0, os.path.abspath('..'))
import numpy as np, matplotlib.pyplot as plt
from src.utils.config import load_config
from src.data.synthetic import SyntheticConfig, generate_vibration_window
from src.physics.bearing_frequencies import SKF_6205, all_fault_frequencies
cfg = SyntheticConfig()
sig = generate_vibration_window('outer', cfg, severity=1.0)
plt.plot(sig[:1000]); plt.title('Outer-race vibration (synthetic)'); plt.show()
```

```python
from src.data.preprocessing import compute_stft_map
m, fax = compute_stft_map(sig, fs=cfg.fs, out_size=64)
plt.imshow(m, origin='lower', aspect='auto', cmap='magma'); plt.title('STFT'); plt.colorbar(); plt.show()
```

```python
spec = np.abs(np.fft.rfft(sig)); freqs = np.fft.rfftfreq(len(sig), 1/cfg.fs)
ff = all_fault_frequencies(cfg.shaft_hz, SKF_6205)
plt.plot(freqs, spec); [plt.axvline(v, color='r', ls='--') for v in ff.values()]
plt.xlim(0,4000); plt.title('FFT with fault-frequency overlays'); plt.show()
```
