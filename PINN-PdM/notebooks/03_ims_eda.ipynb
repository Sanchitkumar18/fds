# 03 — IMS Degradation EDA

Health-index trajectories per bearing unit (or synthetic run-to-failure fallback).

```python
import sys, os
sys.path.insert(0, os.path.abspath('..'))
import numpy as np, matplotlib.pyplot as plt
from src.utils.config import load_config
from src.pipeline import prepare_dataset
cfg = load_config('config/ims.yaml')
proc, _ = prepare_dataset(cfg)
units = np.unique(proc['unit_id'])
print('units', len(units))
```

```python
for u in units[:5]:
    m = proc['unit_id']==u
    o = np.argsort(proc['t'][m])
    plt.plot(proc['t'][m][o], proc['health'][m][o], label=f'unit {u}')
plt.legend(); plt.xlabel('norm time'); plt.ylabel('health'); plt.title('IMS degradation trajectories'); plt.show()
```
