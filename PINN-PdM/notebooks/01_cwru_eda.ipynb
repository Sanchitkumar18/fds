# 01 — CWRU Exploratory Data Analysis

Loads the CWRU processed cache (or synthetic fallback), inspects class/load distribution and STFT maps, and overlays theoretical bearing fault frequencies.

```python
import sys, os
sys.path.insert(0, os.path.abspath('..'))
import numpy as np, matplotlib.pyplot as plt
from src.utils.config import load_config
from src.pipeline import prepare_dataset
cfg = load_config('config/cwru.yaml')
proc, _ = prepare_dataset(cfg)
print('windows', proc['stft'].shape, 'synthetic_fallback', bool(int(proc.get('_synthetic',[1])[0])))
```

```python
classes = list(cfg['cwru']['fault_classes'])
counts = np.bincount(proc['label'], minlength=len(classes))
plt.bar(classes, counts); plt.title('CWRU class distribution'); plt.show()
```

```python
from src.physics.bearing_frequencies import SKF_6205, all_fault_frequencies, cwru_shaft_hz
print('Fault freqs @0HP:', all_fault_frequencies(cwru_shaft_hz(0), SKF_6205))
```

```python
g = int(np.where(proc['label']==1)[0][0])
plt.imshow(proc['stft'][g], origin='lower', aspect='auto', cmap='magma'); plt.title('STFT — inner race'); plt.colorbar(); plt.show()
```
